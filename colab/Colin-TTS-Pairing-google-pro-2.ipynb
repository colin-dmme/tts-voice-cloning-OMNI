Chọn VieNeu v3 (23 preset gồm Ngọc Huyền + clone Profile) hoặc ZeroTTS, rồi chạy cell dưới.
Notebook hiện mã ngắn: mở Studio → Điện toán Colab → Kết nối Colab, nhập mã và kiểm tra đúng worker/model trước khi xác nhận.
Không cần GitHub token hay Secrets. Mã dùng một lần trong 10 phút; phiên worker tối đa 12 giờ, không kéo dài thời hạn runtime Google.
Giữ cell chạy khi dùng; dừng runtime khi xong. Dừng notebook cũ cùng WORKER_ID trước khi ghép lại.
Bạn tự đăng nhập Google/cấp runtime. Chỉ dùng tài nguyên được phép; không dùng nhiều tài khoản để vượt giới hạn, không tự động giữ phiên.


```python
# Không cần Colab Secrets hay GitHub token.
WORKER_ID = 'google-pro-2'
COMPUTE_PROVIDER = 'vieneu_v3' #@param ["vieneu_v3", "zerotts"]
VIENEU_DEVICE = 'cuda' #@param ["cuda", "cpu"]
BROKER_URL = 'https://tts-compute.tainguyenso.online'

from urllib.request import Request, build_opener, HTTPRedirectHandler
class NoRedirect(HTTPRedirectHandler):
    def redirect_request(self, req, fp, code, msg, headers, newurl):
        raise RuntimeError('Không theo chuyển hướng khi tải bootstrap')
_opener = build_opener(NoRedirect())
with _opener.open(Request(BROKER_URL + '/colab/bootstrap.py', headers={'User-Agent': 'Colin-TTS-Compute/0.2'}), timeout=30) as _response:
    _source = _response.read(128 * 1024 + 1)
if len(_source) > 128 * 1024:
    raise RuntimeError('Bootstrap vượt giới hạn')
_namespace = {'__name__': 'colin_pairing_bootstrap'}
exec(compile(_source, '<colin-pairing-bootstrap>', 'exec'), _namespace)
_namespace['run'](BROKER_URL, WORKER_ID, COMPUTE_PROVIDER, 'cpu' if COMPUTE_PROVIDER == 'zerotts' else VIENEU_DEVICE)
```
